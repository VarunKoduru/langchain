```python
def l_d(fn):
  with open(fn,'r',encoding="utf-8") as f:
    return f.read()

text = l_d("/content/sx.txt").replace(".","")
text1 = text.split()


def ch(t,overlap,chunk_size):
  chunks = []
  step = chunk_size - overlap
  for i in range(0,len(t),step):
    temp = " ".join(t[i:i+chunk_size])
    if temp:
      chunks.append(temp)
    if i + chunk_size >= len(t):
      break
  return chunks

res = ch(text1,20,40)
for r in res:
  print(r)

```

```python
import numpy as np
from fastembed import TextEmbedding


def fn(file):
  with open(file,"r",encoding="utf-8") as f:
    text = f.read().replace(".","")
    t1 = text.split()
    return t1

def ch(chunks,ol,cs):
  cx = []
  step = cs -  ol
  for i in range(0,len(chunks),step):
    temp = " ".join(chunks[i:i+cs])
    if temp:
      cx.append(temp)
    if i+cs >= len(chunks):
      break

  return cx

def em(cks):
  em = TextEmbedding()
  vs = np.array(list(em.embed(cks)))
  return vs

def cosine(q_e, v_e):

    q_n = np.linalg.norm(q_e)
    v_n = np.linalg.norm(v_e, axis=1)

    sc = np.dot(v_e, q_e) / (v_n * q_n)

    return sc

def sm(vs,q,k,cx):
  e_model = TextEmbedding()
  qs = np.array(list(e_model.embed([q])))[0]

  scores = cosine(qs,vs)

  top_ind = np.argsort(scores)[::-1][:k]
  res = []
  for i in top_ind:
    res.append({
        "chunk":cx[i],
        "score":float(scores[i])
    })


  return res

def main():
  tokens = fn("/content/sx.txt")
  chunks = ch(tokens,20,80)
  vs = em(chunks)
  q = "What is supervised learning ?"
  result = sm(vs,q,3,chunks)
  for x in result:
    print(x)

main()
```
